```python
from collections import defaultdict

def find_all_paths(variables):
    # Step 1: Create a directed graph
    graph = defaultdict(list)
    n = len(variables)
    
    for i in range(n):
        for j in range(i + 1, n):
            graph[i].append(j)
    
    # Step 2: DFS to find all paths from a given node
    def dfs(node, path):
        paths = []
        for neighbor in graph[node]:
            paths.append(path + [neighbor])  # Add direct path
            paths.extend(dfs(neighbor, path + [neighbor]))  # Recursive paths
        return paths
    
    # Step 3: Find all paths starting from each node
    all_paths = []
    for i in range(n):
        all_paths.extend(dfs(i, [i]))
    
    # Step 4: Convert indices to variable names
    named_paths = [[variables[idx] for idx in path] for path in all_paths]
    return named_paths

# Example variables
variables = ['first_price', 'offer_counteroffer_diff', 'negotiation_length', 'last_price', 'G']

# Find all paths
paths = find_all_paths(variables)

# Print the results
print("All Possible Paths:")
for path in paths:
    print(path)
```

```python
def indirect_dep(a,b,paths,direct):
    a_b_ind_lis =[]
    added = 0
    for path in paths:
        
        if len(path)>2:
            
            if path[0] == a and path[-1] ==b:
                a_b_ind = 1
                for i,j in zip(path[:-1],path[1:]):
                    added = 1
                    
                    a_b_ind *=direct[(i,j)]
        
                a_b_ind_lis.append(a_b_ind)
    if added:
        return sum(a_b_ind_lis)
    else:
        return 0
                    
```

```python
#final_db_distributive_nego_accept['Control']
def calculate_value(offer,cost):
    return offer - cost

def sellers_value(offer,cost_of_production):
    return calculate_value(offer,cost_of_production)
def buyers_value(offer,willingness_to_pay):
    return calculate_value(willingness_to_pay,offer)



def gini_coeff(sellers_value,buyers_value):
    
    return np.absolute( buyers_value - sellers_value)/(buyers_value + sellers_value)
    

cost_of_production = 40
willingness_to_pay = 60
```

```python
def run_reg( data, y = 'U_s',a = 'T1_SV_BD',param_list = ['first_price', 'offer_counteroffer_diff', 'negotiation_length', 'last_price']):

    

    summary_tables = []
    for i in range(len(param_list)+1):
        Y = data[ y]
        X = data[[a] + param_list[:i]]
        # Add a constant term
        X = sm.add_constant(X)
        model = sm.OLS(Y, X).fit()
        display(model.summary2().tables[1].round(3))
        summary_tables.append(model.summary2())
    return summary_tables
```

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from arch.bootstrap import IIDBootstrap

# Example data (serial mediation)
np.random.seed(42)



#['Control', 'T1_SV_BD', 'T2_SD_BV', 'T3_SV_BA', 'T4_SA_BV',
#       'first_price', 'offer_counteroffer_diff', 'negotiation_length',
#       'last_price', 'start_finish_diff', 'U_s', 'U_b', 'G']


#data = pd.DataFrame({'a': a, 'd': d, 'e': e, 'f': f, 'Y': Y})

def calculate_serial_indirect_effects(data, a = 'T1_SV_BD', Y = 'U_s'#, 'U_b', 'G'
                                     ):
    # Model 1: d ~ a

    d, e,f,g = 'first_price', 'offer_counteroffer_diff', 'negotiation_length', 'last_price'

    #print(data[a].mean())

    # T ->ALL ------------------------
    
    # T -> first price
    model_ad = LinearRegression().fit(data[[a]], data[d])
    ad_a = model_ad.coef_[0]

    # T -> OCD
    model_ae = LinearRegression().fit(data[[a]], data[e])
    ae_a = model_ae.coef_[0]

    
    # D -> negotiation_length
    #model_af = LinearRegression().fit(data[[a]], data[f])
    #af_a = model_af.coef_[0]

    # D -> LP
    #model_ag = LinearRegression().fit(data[[a]], data[g])
    #ag_a = model_ag.coef_[0]
    
    # T -> Y 
    model_aY = LinearRegression().fit(data[[a]], data[Y])
    aY_a = model_aY.coef_[0]

    





    
    # A D ->ALL ------------------------

    # T + FO ~ OFD
    model_ade = LinearRegression().fit(data[[a,d]], data[e])
    ade_a = model_ade.coef_[0]
    ade_d = model_ade.coef_[1]

    # T + FO ~ Y
    model_adY = LinearRegression().fit(data[[a,d]], data[Y])
    adY_a = model_adY.coef_[0]
    adY_d = model_adY.coef_[1]


    # A D E ->ALL ------------------------

    # T + FO + OCD ~ negotiation_length
    model_adef = LinearRegression().fit(data[[a,d,e ]], data[f])
    adef_a = model_adef.coef_[0]
    adef_d = model_adef.coef_[1]
    adef_e = model_adef.coef_[2]

    # T + FO + OCD ~ Y
    model_adeY = LinearRegression().fit(data[[a,d,e]], data[Y])
    adeY_a = model_adeY.coef_[0]
    adeY_d = model_adeY.coef_[1]  
    adeY_e = model_adeY.coef_[2]  
    

    # A D E F->ALL ------------------------

    # T + FO + OCD + negotiation_length ~ last_price
    model_adefg = LinearRegression().fit(data[[a,d,e,f]], data[g])
    adefg_a = model_adefg.coef_[0]
    adefg_d = model_adefg.coef_[1]
    adefg_e = model_adefg.coef_[2]
    adefg_f = model_adefg.coef_[3]

    # T + FO + OCD + negotiation_length  ~ Y
    model_adefY = LinearRegression().fit(data[[a,d,e,f]], data[Y])
    adefY_a = model_adefY.coef_[0]
    adefY_d = model_adefY.coef_[1]  
    adefY_e = model_adefY.coef_[2] 
    adefY_f = model_adefY.coef_[3] 


    # A D E F G->ALL ------------------------

    # T + FO + OCD + negotiation_length + last_price ~ Y
    model_adefgY = LinearRegression().fit(data[[a,d,e,f,g]], data[Y])
    adefgY_a = model_adefgY.coef_[0]
    adefgY_d = model_adefgY.coef_[1]  
    adefgY_e = model_adefgY.coef_[2] 
    adefgY_f = model_adefgY.coef_[3] 
    adefgY_g = model_adefgY.coef_[4] 
        

    # Calculate specific indirect effects
    indirect_a_d_Y = ad_a * adY_d
    indirect_a_e_Y = ae_a * adeY_e
    indirect_d_e_Y = ade_d * adeY_e # e ~ a +d , Y ~ a + d + e

    indirect_e_f_Y = adef_e * adefY_f
    indirect_f_g_Y = adefg_f * adefgY_g

    indirect_a_g_Y = adefg_a*adefgY_g
    indirect_d_g_Y = adefg_d*adefgY_g
    indirect_e_g_Y = adefg_e*adefgY_g




    
    return np.array([

    indirect_a_d_Y
    ,indirect_a_e_Y
    ,indirect_d_e_Y
    ,indirect_e_f_Y
    ,indirect_f_g_Y
        
    ,indirect_a_g_Y
    ,indirect_d_g_Y
    ,indirect_e_g_Y
        
        
    ]) #Return as a vector


# Bootstrapping

results_dict = dict({})

for a in ['T1_SV_BD','T2_SD_BV', 'T3_SV_BA', 'T4_SA_BV']:
    no_chosen_columns = list({ 'T1_SV_BD','T2_SD_BV', 'T3_SV_BA', 'T4_SA_BV'} - {a})
        
    corre_df_2 = corre_df[ corre_df[no_chosen_columns].sum(axis =1) == 0]
    #rune_1_results =  run_reg( corre_df_2, y = 'G',a = 'T1_SV_BD')
    
    for Y in ['last_price']:#:['U_b','U_s','RFD']:


        
        print('target',Y, ' treatment',a, '--------------------' )
        
        
        
        
        bs = IIDBootstrap(corre_df_2)
        bs.seed(123)
        ci = bs.conf_int(calculate_serial_indirect_effects, reps=1000, method='bca', size=0.95, extra_kwargs =dict(a=a,Y=Y))
        
        # Get point estimates from the original data
        point_estimates = calculate_serial_indirect_effects(corre_df_2)

        results_dict[(Y,a)] = {'ci':ci, 'point_estimates':point_estimates,
                              'main_effects' : run_reg( corre_df_2, y = Y,a = a,
        param_list = ['first_price', 'offer_counteroffer_diff', 'negotiation_length','interaction_L_LP','last_price'])


                               
                              }

        d,e,f,g = 'first_price', 'offer_counteroffer_diff', 'negotiation_length','last_price'
        
        # Print results (with interpretation)
        indirect_effect_names = [
            f"{a} -> {d} -> {Y}",
            f"{a} -> {e} -> {Y}",
            f"{d} -> {e} -> {Y}",
            f"{e} -> {f} -> {Y}",
            f"{f} -> {g} -> {Y}",

            f"{a} -> {g} -> {Y}",
            f"{d} -> {g} -> {Y}",
            f"{e} -> {g} -> {Y}",
                        
        ]
        
        for i, name in enumerate(indirect_effect_names):
            print(f"Indirect Effect: {name}")
            print(f"  Point Estimate: {point_estimates[i]:.3f}")
            print(f"  95% CI: ({ci[0, i]:.3f}, {ci[1, i]:.3f})")
            if ci[0, i] <= 0 <= ci[1, i]:
                print("  NOT Statistically Significant")
            else:
                print("  Statistically Significant")
            print("-" * 30)
# a is the treatement indicator, Y is the as output.
#d, e,f,g = 'first_price', 'offer_counteroffer_diff', 'negotiation_length', 'last_price'
```

```python
data = corre_df[(corre_df.Control ==1) | (corre_df.T2_SD_BV ==1)]
# Assume 'data' is your pandas DataFrame containing columns:
# 'Y', 'T' (categorical or dummy codes), 'a', 'OCD', 'NL'

# --- Step 1: Prepare Variables (Mean-Centering) ---
data['OCD_c'] = data['offer_counteroffer_diff'] - data['offer_counteroffer_diff'].mean()
data['NL_c'] = data['negotiation_length'] - data['negotiation_length'].mean()

# Create interaction term using centered variables
data['OCD_x_NL'] = data['offer_counteroffer_diff'] * data['negotiation_length']
```

```python

# Define Treatment Dummies (Example: Assuming T has levels T1, T2, T3, T4 and Control=C)
# You might already have these, or create them like this:
#data = pd.get_dummies(data, columns=['T'], prefix='T', drop_first=True)
# This creates T_T1, T_T2, T_T3, T_T4 where Control is the reference (all dummies = 0)
# Adjust the prefix and levels based on your actual 'T' column

# --- Step 2 & 3: Run Moderated Regression and Examine Interaction ---

# Example: Running for T1 vs. Control
# Create a subset of the data for the specific comparison
#data_t1_vs_c = data[(data['T_T1'] == 1) | ((data['T_T1'] == 0) & (data['T_T2'] == 0) & (data['T_T3'] == 0) & (data['T_T4'] == 0))]
# Make sure the condition correctly identifies T1 and Control based on your dummies

data_t1_vs_c = data
data_t1_vs_c['T_T1'] = data_t1_vs_c.T2_SD_BV

#data_t1_vs_c['Y'] =data_t1_vs_c.last_price

# Define the model formula (using centered variables)
# We use 'T_T1' as the predictor representing the Treatment effect vs Control
formula_t1 = 'last_price ~ T_T1 + first_price + offer_counteroffer_diff + negotiation_length + OCD_x_NL'

# Fit the OLS model using statsmodels
model_t1 = smf.ols(formula_t1, data=data_t1_vs_c).fit()

# Print the summary to see coefficients, p-values, CIs etc.
print("\n--- Moderation Results for T1 vs. Control ---")
print(model_t1.summary())

# Get the specific coefficient and p-value for the interaction term
interaction_coef_t1 = model_t1.params['OCD_x_NL']
interaction_pvalue_t1 = model_t1.pvalues['OCD_x_NL']
print(f"\nInteraction Term (OCD_x_NL) Coefficient: {interaction_coef_t1:.4f}")
print(f"Interaction Term (OCD_x_NL) P-value: {interaction_pvalue_t1:.4f}")
```

```python
# --- Step 4: Simple Slopes Analysis (if interaction is significant) ---

#offer_counteroffer_diff
#negotiation_length

# Define levels for the moderator (NL)
nl_mean = data['negotiation_length'].mean() # Or use 0 for NL_c
nl_std = data['negotiation_length'].std()
nl_low = nl_mean - nl_std   # Low NL (-1 SD)
nl_high = nl_mean + nl_std  # High NL (+1 SD)

# Calculate corresponding centered values
nl_c_low = nl_low - nl_mean   # This is simply -nl_std
nl_c_high = nl_high - nl_mean # This is simply +nl_std
nl_c_mean = 0                 # Mean of centered NL is 0

# --- Simple Slopes for T1 vs. Control (using model_t1 results) ---
print("\n--- Simple Slopes for T1 vs. Control ---")

# Coefficients from model_t1
beta_ocd_c = model_t1.params['offer_counteroffer_diff']
beta_interaction = model_t1.params['OCD_x_NL']

# Calculate simple slopes
slope_low_nl_t1 = beta_ocd_c + beta_interaction * nl_c_low
slope_mean_nl_t1 = beta_ocd_c + beta_interaction * nl_c_mean # Should equal beta_ocd_c
slope_high_nl_t1 = beta_ocd_c + beta_interaction * nl_c_high

print(f"Simple slope of OCD at Low NL (-1 SD): {slope_low_nl_t1:.4f}")
print(f"Simple slope of OCD at Mean NL: {slope_mean_nl_t1:.4f}")
print(f"Simple slope of OCD at High NL (+1 SD): {slope_high_nl_t1:.4f}")
```

```python
# --- Testing Significance of Simple Slopes (More Advanced) ---
# Statsmodels doesn't have a built-in function as simple as PROCESS or R's
# 'interactions' package for directly testing simple slopes significance easily.
# You typically need to use matrix methods, `lincom`-like tests, or re-centering.

# Method 1: Re-centering NL (Easier to implement)
# Test slope at Low NL: Center NL around the low value
data_t1_vs_c['NL_c_low_centered'] = data_t1_vs_c['negotiation_length'] - nl_c_low
data_t1_vs_c['OCD_x_NL_low_centered'] = data_t1_vs_c['offer_counteroffer_diff'] * data_t1_vs_c['NL_c_low_centered']
formula_t1_low_nl = 'last_price ~ T_T1 + first_price + offer_counteroffer_diff + NL_c_low_centered + OCD_x_NL_low_centered'
model_t1_low_nl = smf.ols(formula_t1_low_nl, data=data_t1_vs_c).fit()
print("\nTesting slope at Low NL:")
# The coefficient (and p-value) for 'OCD_c' in THIS model is the simple slope at Low NL
print(model_t1_low_nl.summary().tables[1])
print(f"P-value for slope at Low NL: {model_t1_low_nl.pvalues['offer_counteroffer_diff']:.4f}")
```

### T -> b -> Y | a {indirect through Coeff(b)*Coeff(a in b ~ T + a)*Coeff(T in a ~ T) }
- controling for a

```python
import statsmodels.formula.api as smf
import pandas as pd
import numpy as np
from arch.bootstrap import IIDBootstrap # Using arch's IIDBootstrap

# Assume 'data' is your pandas DataFrame containing columns:
# 'Y', 'T' (categorical), 'a', 'OCD', 'NL'

# --- Optional: Center Variables ---
# data['a_c'] = data['a'] - data['a'].mean()
# data['OCD_c'] = data['OCD'] - data['OCD'].mean()
# data['NL_c'] = data['NL'] - data['NL'].mean()
# data['OCD_x_NL'] = data['OCD_c'] * data['NL_c'] # If needed for other analyses

# Define Treatment Groups and Control
treatment_groups = ['T1_SV_BD', 'T2_SD_BV', 'T3_SV_BA', 'T4_SA_BV'] # Adjust names as needed
control_group = 'Control_SV_BV' # Adjust control group name as needed

# --- Function to calculate the two specific indirect effects ---
# This function now takes the resampled data directly
def calculate_ocd_indirect_effects_arch(data, treatment_dummy_col='T1_SV_BD',a = 'first_price',OCD = 'offer_counteroffer_diff',Y = 'last_price'):
    
    """
    Calculates the two indirect effects involving OCD from a data sample.

    Args:
        df_sample (pd.DataFrame): A (potentially resampled) DataFrame.
        treatment_dummy_col (str): The name of the 0/1 treatment dummy column.

    Returns:
        np.array: [indirect_effect_1, indirect_effect_2] or [np.nan, np.nan].
    """
    try:
        # Check if dummy column exists, create if not (needed for first call)
        #if treatment_dummy_col not in df_sample.columns:
        #     treatment_name = treatment_dummy_col.split('_', 1)[1] # Extract T1, T2 etc.
        #     df_sample[treatment_dummy_col] = df_sample['T'].apply(lambda x: 1 if x == treatment_name else 0)

        # Fit required models using formula API

        model_a = smf.ols(f'{a} ~ {treatment_dummy_col}', data=data).fit()
        model_ocd = smf.ols(f'{OCD} ~ {treatment_dummy_col} + {a}', data=data).fit()
        model_y = smf.ols(f'{Y} ~ {treatment_dummy_col} + {a} + {OCD}', data=data).fit()

        # Extract coefficients (handle potential missing keys if model failed)
        coef_T_in_a = model_a.params.get(treatment_dummy_col, np.nan)
        coef_T_in_OCD = model_ocd.params.get(treatment_dummy_col, np.nan)
        coef_a_in_OCD = model_ocd.params.get(a, np.nan)
        coef_OCD_in_Y = model_y.params.get(OCD, np.nan)

        if np.isnan([coef_T_in_a, coef_T_in_OCD, coef_a_in_OCD, coef_OCD_in_Y]).any():
             return np.array([np.nan, np.nan])

        # Calculate indirect effects
        indirect_effect_1 = coef_T_in_OCD * coef_OCD_in_Y # T -> OCD -> Y | a
        indirect_effect_2 = coef_T_in_a * coef_a_in_OCD * coef_OCD_in_Y # T -> a -> OCD -> Y

        return np.array([indirect_effect_1, indirect_effect_2])
    except Exception as e:
        print(f"Error calculating indirect effects in sample: {e}")
        return np.array([np.nan, np.nan])

# --- Loop through each Treatment vs. Control comparison ---

n_bootstrap_samples = 5000
ci_method = 'bca' # Bias-corrected and accelerated
random_seed = 42

treatment_dummy_col='T1_SV_BD'
a = 'first_price'
OCD = 'offer_counteroffer_diff'
Y = 'last_price'


all_results = {} # Dictionary to store results

for treatment_name in treatment_groups:
    #print(f"\n{'='*20} ANALYSIS: {treatment_name} vs. {control_group} {'='*20}")

    # Create subset for the current comparison
    #data_subset = data[data['T'].isin([treatment_name, control_group])].copy()
    data_subset = corre_df[(corre_df.Control ==1) | (corre_df[treatment_name] ==1)].copy()

    # Define the specific dummy column name for this treatment
    #treatment_dummy_col = f"T_{treatment_name}" # e.g., T_T1_SV_BD
    treatment_dummy_col = treatment_name
    a = 'first_price'
    OCD = 'offer_counteroffer_diff'
    Y = 'last_price'

    # Create the 0/1 dummy variable within the subset
    # Do this *before* bootstrapping so it's in the data passed to IIDBootstrap
    #data_subset[treatment_dummy_col] = data_subset['T'].apply(lambda x: 1 if x == treatment_name else 0)

    # --- Perform Bootstrapping with IIDBootstrap ---
    try:
        bs = IIDBootstrap(data_subset[[Y, treatment_dummy_col, a, OCD]]) # Pass necessary columns
        bs.seed(random_seed)

        # Calculate confidence intervals for the function
        # Use a lambda to pass the specific treatment_dummy_col name
        #ci = bs.conf_int(
        #    lambda df: calculate_ocd_indirect_effects_arch(df, treatment_dummy_col = treatment_dummy_col),
        #    reps=n_bootstrap_samples,
        #    method=ci_method
        #)

        ci = bs.conf_int(calculate_ocd_indirect_effects_arch,
                         reps=n_bootstrap_samples, method=ci_method, size=0.95, 
                         extra_kwargs =dict(treatment_dummy_col = treatment_dummy_col))
        
        # Get point estimates from the original data
        #point_estimates = calculate_serial_indirect_effects(corre_df_2,treatment_dummy_col = treatment_dummy_col)
        

        # Calculate point estimates using the original subset
        point_estimates = calculate_ocd_indirect_effects_arch(data_subset, 
                                                              treatment_dummy_col= treatment_dummy_col)

        # --- Store and Print Results ---
        results_dict = {
            'IE1_T_OCD_Y': {
                'Estimate': point_estimates[0],
                'CI_Low': ci[0, 0],
                'CI_High': ci[1, 0],
                'Significant': not (ci[0, 0] <= 0 <= ci[1, 0])
            },
            'IE2_T_a_OCD_Y': {
                'Estimate': point_estimates[1],
                'CI_Low': ci[0, 1],
                'CI_High': ci[1, 1],
                'Significant': not (ci[0, 1] <= 0 <= ci[1, 1])
            }
        }
        all_results[treatment_name] = results_dict

        print(f"Indirect Effect 1 (T -> OCD -> Y | a):")
        print(f"  Point Estimate: {results_dict['IE1_T_OCD_Y']['Estimate']:.4f}")
        print(f"  95% {ci_method.upper()} CI: ({results_dict['IE1_T_OCD_Y']['CI_Low']:.4f}, {results_dict['IE1_T_OCD_Y']['CI_High']:.4f})")
        print(f"  Significant: {results_dict['IE1_T_OCD_Y']['Significant']}")

        print(f"\nIndirect Effect 2 (T -> a -> OCD -> Y):")
        print(f"  Point Estimate: {results_dict['IE2_T_a_OCD_Y']['Estimate']:.4f}")
        print(f"  95% {ci_method.upper()} CI: ({results_dict['IE2_T_a_OCD_Y']['CI_Low']:.4f}, {results_dict['IE2_T_a_OCD_Y']['CI_High']:.4f})")
        print(f"  Significant: {results_dict['IE2_T_a_OCD_Y']['Significant']}")

    except Exception as e:
        print(f"Could not run bootstrapping for {treatment_name}: {e}")
        all_results[treatment_name] = None # Indicate failure

    #print(f"\n{'='*20} END ANALYSIS: {treatment_name} vs. {control_group} {'='*20}")

# --- Optional: Display results summary ---
# print("\n--- Summary of Indirect Effects ---")
# for treatment, results in all_results.items():
#     if results:
#         print(f"\nTreatment: {treatment}")
#         print(f"  IE1 (T->OCD->Y|a): Est={results['IE1_T_OCD_Y']['Estimate']:.3f}, CI=[{results['IE1_T_OCD_Y']['CI_Low']:.3f}, {results['IE1_T_OCD_Y']['CI_High']:.3f}], Sig={results['IE1_T_OCD_Y']['Significant']}")
#         print(f"  IE2 (T->a->OCD->Y): Est={results['IE2_T_a_OCD_Y']['Estimate']:.3f}, CI=[{results['IE2_T_a_OCD_Y']['CI_Low']:.3f}, {results['IE2_T_a_OCD_Y']['CI_High']:.3f}], Sig={results['IE2_T_a_OCD_Y']['Significant']}")
#     else:
#         print(f"\nTreatment: {treatment} - Analysis Failed")
```

```python
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf
from arch.bootstrap import IIDBootstrap
import warnings

# --- 1. Configuration ---
# Assume 'data' is your pandas DataFrame
x_var = 'T2_SD_BV'         # Independent Variable
m1_var = 'first_price'      # First Mediator
m2_var = 'Counter_offer'    # Second Mediator
y_var = 'last_price'        # Dependent Variable

# Define your control variables (replace or set to [] if none)
# Ensure these are columns in your 'data' DataFrame
control_vars = []#'Control1', 'Control2'] # EXAMPLE control variables

# Construct formula snippets
x_term = x_var
m1_term = m1_var
m2_term = m2_var
y_term = y_var


data = corre_df[(corre_df.Control ==1) | (corre_df[x_var] ==1)]

# Controls string (add controls only if they exist)
if control_vars:
    # Check if all control variables are actually in the DataFrame columns
    missing_controls = [c for c in control_vars if c not in data.columns]
    if missing_controls:
        print(f"Warning: Control variables {missing_controls} not found in DataFrame columns. Removing them.")
        control_vars = [c for c in control_vars if c in data.columns]

    if control_vars: # Check again in case all were missing
         controls_str = " + " + " + ".join(control_vars)
    else:
        controls_str = ""
else:
    controls_str = ""
```

```python
# --- 2. Define Model Formulas ---
# Model for M1 (first_price)
formula_m1 = f"{m1_term} ~ {x_term}{controls_str}"
# Model for M2 (counter_offer), predicted by X and M1
formula_m2 = f"{m2_term} ~ {x_term} + {m1_term}{controls_str}"
# Model for Y (deal_price), predicted by X, M1, and M2
formula_y = f"{y_term} ~ {x_term} + {m1_term} + {m2_term}{controls_str}"

print("--- Formulas ---")
print(f"M1 ({m1_var}): {formula_m1}")
print(f"M2 ({m2_var}): {formula_m2}")
print(f"Y ({y_var}):  {formula_y}")
print("----------------")
```

```python
# --- 3. Fit models on original data & get point estimates ---
if True: #try:
    model_m1_fit = smf.ols(formula=formula_m1, data=data).fit()
    model_m2_fit = smf.ols(formula=formula_m2, data=data).fit()
    model_y_fit = smf.ols(formula=formula_y, data=data).fit()

    print("\n--- Model Summaries (Original Data) ---")
    print(f"\n--- Outcome: {m1_var} ---")
    print(model_m1_fit.summary())#yname=m1_var, xname=[x_term] + control_vars))
    print(f"\n--- Outcome: {m2_var} ---")
    print(model_m2_fit.summary())#yname=m2_var, xname=[x_term, m1_term] + control_vars))
    print(f"\n--- Outcome: {y_var} ---")
    print(model_y_fit.summary())#yname=y_var, xname=[x_term, m1_term, m2_term] + control_vars))
    print("---------------------------------------")

    # Extract key coefficients for the serial path: X -> M1 -> M2 -> Y
    params_m1 = model_m1_fit.params
    params_m2 = model_m2_fit.params
    params_y = model_y_fit.params

    # Path coefficients (using common notation where possible)
    a1 = params_m1.get(x_term, 0)      # X -> M1 path
    d21 = params_m2.get(m1_term, 0)    # M1 -> M2 path (controlling for X)
    b2 = params_y.get(m2_term, 0)      # M2 -> Y path (controlling for X, M1)

    # Calculate the point estimate of the specific serial indirect effect
    serial_indirect_effect_point_estimate = a1 * d21 * b2

    print(f"\n--- Point Estimate of Serial Indirect Effect ---")
    print(f"Path T -> {m1_var} (a1): {a1:.4f}")
    print(f"Path {m1_var} -> {m2_var} (d21): {d21:.4f}")
    print(f"Path {m2_var} -> {y_var} (b2): {b2:.4f}")
    print(f"Serial Indirect Effect ({x_var}->{m1_var}->{m2_var}->{y_var}): {serial_indirect_effect_point_estimate:.4f}")
    print("-------------------------------------------------")

#except Exception as e:
#    print(f"Error fitting initial models: {e}")
#    serial_indirect_effect_point_estimate = np.nan # Set to NaN if initial fit fails

```

```python
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf
from arch.bootstrap import IIDBootstrap
import warnings



# --- Configuration ---
# Assume 'data' is your pandas DataFrame
i = x_var = 'T2_SD_BV'


data = corre_df[(corre_df.Control ==1) | (corre_df[i] ==1)]


m1_var = 'first_price'
m2_var = 'offer_counteroffer_diff'
y_var = 'last_price'
w_var = 'negotiation_length'

# Define your control variables (replace or set to [])
# Ensure these are also columns in your 'data' DataFrame
control_vars = []#'Your_Control1', 'Your_Control2']

# Construct formula snippets
x_term = x_var
m1_term = m1_var
m2_term = m2_var
y_term = y_var
w_term = w_var
interaction_term = f"{m2_term}:{w_term}" # Colon for interaction in statsmodels

# Controls string (ensure controls are added only if they exist)
if control_vars:
    controls_str = " + " + " + ".join(control_vars)
else:
    controls_str = ""

# --- Define Model Formulas ---
formula_m1 = f"{m1_term} ~ {x_term}{controls_str}"
formula_m2 = f"{m2_term} ~ {x_term} + {m1_term}{controls_str}"
formula_y = f"{y_term} ~ {x_term} + {m1_term} + {m2_term} + {w_term} + {interaction_term}{controls_str}"

print("--- Formulas ---")
print(f"M1: {formula_m1}")
print(f"M2: {formula_m2}")
print(f"Y:  {formula_y}")
print("----------------")

# --- Part 1: Fit models on original data & get point estimates ---
# (Same as before to get the point estimate and check the M2*W interaction)
try:
    model_m1_fit = smf.ols(formula=formula_m1, data=data).fit()
    model_m2_fit = smf.ols(formula=formula_m2, data=data).fit()
    model_y_fit = smf.ols(formula=formula_y, data=data).fit()

    print("\n--- Results for Y Model (Check Interaction Term) ---")
    print(model_y_fit.summary(yname=y_var))
    print("----------------------------------------------------")

    params_m1 = model_m1_fit.params
    params_m2 = model_m2_fit.params
    params_y = model_y_fit.params

    a1 = params_m1.get(x_term, 0)
    d21 = params_m2.get(m1_term, 0)
    b3 = params_y.get(interaction_term, 0)

    index_point_estimate_serial = a1 * d21 * b3

    print(f"\n--- Point Estimate ---")
    print(f"Coefficient for M2*W interaction ({interaction_term}): {b3:.4f}")
    print(f"  (Check p-value/CI for this in the summary above for significance of M2->Y moderation)")
    print(f"Index of Moderated Mediation (X->M1->M2->Y path) Point Estimate: {index_point_estimate_serial:.4f}")
    print("--------------------")

except Exception as e:
    print(f"Error fitting initial models: {e}")
    index_point_estimate_serial = np.nan
```

```python
# --- Part 2: Define the function to compute the statistic ---
def compute_moderated_mediation_index(boot_data):
    """
    Fits the 3 models and computes the serial moderated mediation index (a1*d21*b3).
    Takes a DataFrame (a bootstrap sample) as input.
    Returns the index value or np.nan if models fail.
    """
    try:
        # Fit models on the bootstrap sample
        boot_m1_fit = smf.ols(formula=formula_m1, data=boot_data).fit()
        boot_m2_fit = smf.ols(formula=formula_m2, data=boot_data).fit()
        boot_y_fit = smf.ols(formula=formula_y, data=boot_data).fit()

        # Extract coefficients
        params_m1_boot = boot_m1_fit.params
        params_m2_boot = boot_m2_fit.params
        params_y_boot = boot_y_fit.params

        a1_boot = params_m1_boot.get(x_term, 0)
        d21_boot = params_m2_boot.get(m1_term, 0)
        b3_boot = params_y_boot.get(interaction_term, 0) # M2*W interaction coef

        # Calculate index
        index_boot_serial = a1_boot * d21_boot * b3_boot
        return index_boot_serial

    except Exception:
        # Return NaN if any model fails to fit on this bootstrap sample
        return np.nan

# --- Part 3: Run Bootstrap using arch.bootstrap ---
n_boot = 5000 # Number of bootstrap samples

print(f"\n--- Running Bootstrap ({n_boot} samples) using arch.bootstrap ---")
warnings.filterwarnings("ignore") # Suppress potential warnings during bootstrap

# Pass the entire DataFrame to IIDBootstrap
# The compute_moderated_mediation_index function expects a DataFrame
bs = IIDBootstrap(data, seed=42) # Set seed for reproducibility

# Compute confidence intervals using the percentile method
# Pass the function that calculates our statistic
ci_results = bs.conf_int(compute_moderated_mediation_index,
                           reps=n_boot,
                           method='percentile',
                           size=0.95, # For 95% CI
                           tail='two') # Two-tailed CI

# Get standard errors (optional, but good practice)
#se_results = bs.std_err(compute_moderated_mediation_index, reps=n_boot)

# Get point estimates from the original data
index_point_estimate_serial = compute_moderated_mediation_index(data)

        #results_dict[(Y,a)] = {'ci':ci, 'point_estimates':point_estimates,

warnings.filterwarnings("default") # Turn warnings back on
print("--- Bootstrap Complete ---")

# --- Part 4: Display Bootstrap Results ---
# The conf_int result is usually a 1x2 array/list [lower, upper]
# The std_err result is usually a scalar

lower_ci = ci_results[ 0,0]
upper_ci = ci_results[ 1,0]
#index_se_boot = se_results[0] # Assuming it returns a single value

print("\n--- Bootstrap Results for Index of Moderated Mediation (X->M1->M2->Y) ---")
print(f"Point Estimate (from original data): {index_point_estimate_serial:.4f}")
#print(f"Bootstrap SE: {index_se_boot:.4f}")
print(f"95% Percentile CI: [{lower_ci:.4f}, {upper_ci:.4f}]")
print("-----------------------------------------------------------------------")

# Interpretation Aid
if lower_ci < 0 < upper_ci:
    print("Interpretation: The 95% CI includes zero, suggesting the moderated mediation effect for the X->M1->M2->Y path is NOT statistically significant.")
else:
    print("Interpretation: The 95% CI does NOT include zero, suggesting the moderated mediation effect for the X->M1->M2->Y path IS statistically significant.")

# Optional: Display full summary from arch bootstrap object
# print("\n--- ARCH Bootstrap Summary Object ---")
# print(bs.summary(compute_moderated_mediation_index, reps=n_boot))
# print("------------------------------------")


```
